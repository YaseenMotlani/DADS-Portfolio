```python
import pandas as pd
```

```python
movies = pd.DataFrame({
    "Genre": ["Drama", "Action", "Comedy"]
}, index=["Dangal", "Jawan", "3 Idiots"])

print(movies)
```

```python
ratings = pd.DataFrame({
    "Rating": [8.4, 7.0, 8.4]
}, index=["Dangal", "Jawan", "3 Idiots"])

print(ratings)
```

```python
# left join
```

```python
print(movies.join(ratings,how='left'))
```

```python
print(movies.join(ratings, how='right'))
```

```python
print(movies.join(ratings, how='inner'))
```
