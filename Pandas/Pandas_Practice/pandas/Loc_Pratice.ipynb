LOC & ILOC LOGIC BUILDING

```python
import pandas as pd
```

```python
data = {
    "Patient_ID": ["P101", "P102", "P103", "P104", "P105", "P106", "P107", "P108", "P109", "P110"],
    "Age": [45, 30, 55, 62, 28, 41, 50, 33, 47, 39],
    "City": ["Hyderabad", "Warangal", "Chennai", "Hyderabad", "Mumbai", "Warangal", "Hyderabad", "Chennai", "Hydearabd", "Mumbai"],
    "Surgery_Type": ["Cardiac", "Ortha", "Neuro", "Cardiac", "Ortho", "Neuro", "Cardiac", "Ortho", "Neuro", "Cardiac"],
    "Admission_Type": ["Planned", "Emergency", "Planned", "Planned", "Emergency", "Planned", "Emergency", "Planned","Planned", "Emergency"],
    "Operation_Duration_Min": [120, 60, 150, 90, 45, 110, 130, 70, 160, 50]
}
```

```python
df=pd.DataFrame(data)
```

```python
print(df)
```

Q1. Using loc, display only the patients whose City is
"Hyderabad" OR "Warangal". (Display only:Patient_ID,Age,City,Surgery_Type)


```python
df.loc[(df["City"] == "Hyderabad") | (df["City"] == "Warangal"),[ "Patient_ID", "Age", "City", "Surgery_Type"]]
```

Q2. Using iloc, display:• Rows 2 to 7,• Only columns 1, 3 and 5
 Remember: iloc uses POSITION, not column names.

```python
df.iloc[2:8, [1, 3, 5]]
```

Q3.Using loc, display patients whose:• Age is greater than 40 AND • Admission_Type is "Planned"
Display only: Patient_ID, Age, Surgery_Type, Admission_Type
.


```python
df.loc[(df["Age"] > 40) & (df["Admission_Type"] == "Planned"), ["Patient_ID", "Age", "Surgery_Type", "Admission_Type"] ]
```

Q4. Using loc, find patients from Hyderabad whose Operation_Duration_Min is greater than 100.
Display only: Patient_ID, City, Operation_Duration_Min

```python
df.loc[(df["City"] == "Hyderabad") & (df["Operation_Duration_Min"] > 100), ["Patient_ID", "City", "Operation_Duration_Min"] ]
```

Q5. Using iloc, display:• Every alternate row starting from the first row, • Only the first 4 columns
Think about: row slicing + column slicing

```python
df.iloc[::2, :4]
```
