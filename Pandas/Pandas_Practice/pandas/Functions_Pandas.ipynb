```python

import pandas as pd
```

```python
data = {
    "ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    "Name": [
        "Aman", "Priya", "Rahul", "Sneha", "Arjun",
        "Neha", "Rohit", "Sara", "Vikas", "Anjali"
    ],

    "Age": [21, 22, None, 20, 23, 21, None, 24, 22, 25],

    "City": [
        "Nagpur", "Pune", "Nagpur", "Mumbai", "Pune",
        "Nagpur", "Delhi", "Pune", None , "Nagpur"
    ],

    "Course": [
        "Python", "Data Science", "Python", "SQL", "Python",
        "Power BI", "Data Science", "Python", "SQL", "Power BI"
    ],

    "Python_Marks": [
        85, 78, 92, None, 67,
        88, 75, 95, 54, None
    ],

    "SQL_Marks": [
        78, 85, None, 72, 65,
        88, 70, None, 49, 91
    ],

    "Attendance": [
        92, 85, 96, 78, None,
        88, 65, 97, 59, 90
    ],

    "Fees": [
        50000, 45000, 50000, None, 40000,
        50000, 35000, 50000, None, 40000
    ],

    "Placement_Status": [
        "Placed", "Not Placed", "Placed", "Not Placed", None,
        "Placed", "Not Placed", "Placed", "Not Placed", "Placed"
    ]
}
```

```python
df = pd.DataFrame(data)
```

1) isnull()
`isnull()` is used to check whether values are missing in a DataFrame.


True → Value is missing, 
False → Value is available

```python
df.isnull()
```

```python
df["Age"].isnull()
```

```python
df["Age"].isnull().sum()
```

isnull()       → Where are the missing values?  
isnull().sum() → How many missing values are there?

3) unique()  
unique() returns all unique/different values from a column.

```python
df["City"].unique()
```

4. nunique()  
nunique() returns the number of unique/different values.

```python
df["City"].nunique()
```

unique()  → Shows unique values  
nunique() → Counts unique values

5. value_counts()  
value_counts() counts how many times each value occurs.

```python
df["City"].value_counts()
```

6. std()  
std() stands for Standard Deviation.  
It tells us how much the values are spread out around the mean.

```python
df["Age"].std()
```
