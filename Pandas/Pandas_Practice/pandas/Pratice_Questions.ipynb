```python
import pandas as pd #library ko import karte hain, pd uska short naam hai
```

```python
data = {
    "Name" : ["Aman", "Sara", "Rohan", "Yash", "Neha"],
    "City" : ["Pune", "Mumbai", "Pune","Napur", "Delhi"],
    "Python" : [85, 95, 72, 88, 60],
    "SQL" : [80, 70, 90, 65, 75],
    "Attendance" : [90, 75, 85, 95, 60]
}
```

```python
df = pd.DataFrame(data)  #dictionary se table banata hai
```

```python
print(df)
```

```python
# Q1. Highest Python marks
df["Python"].max()
```

```python
# Q2. Highest Python marks wale student ka naam
df.loc[df["Python"].idxmax(), "Name"]
```

Q3. Find the highest SQL marks.

```python
df["SQL"].max()
```

Q4. Find the student name who scored the highest SQL marks.

```python
df.loc[df["SQL"].idxmax(), "Name"]
```

 Q5.Find the lowest Python marks.

```python
df["Python"].min()
```

Q6. Find the student name who has the lowest Python marks.

```python
df.loc[df["Python"].idxmin(), "Name"]
```

Q7. Find the average Python marks.

```python
df["Python"].median()
```

Q8. Find the average SQL marks.

```python
df["SQL"].median()
```

Q9. Find students whose Python marks are greater than 80.

```python
df.loc[df["Python"] > 80 , "Name" ]
```

Q10. Find students from Pune.

```python
df.loc[df["City"]== "Pune", "Name"]
```
