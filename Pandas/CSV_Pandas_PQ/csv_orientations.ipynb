gp-genre avg rating -----imdb_rating-----mean max---------min--------avg runtime-------mean

```python
csv_movies.groupby("Genre").agg(AVG_Rating=("IMDB_Rating","mean"), Max_Rating = (("IMDB_Rating", "max")), Min_Rating = (("IMDB_Rating", "min")), Avg_Runtime= (("Runtime", "mean")))
```
