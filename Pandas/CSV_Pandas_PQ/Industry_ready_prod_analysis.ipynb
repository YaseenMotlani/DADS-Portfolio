```python
# 1 Audience Engagement
# Which Movies have the strongest audince engaegement singal in the
csv.nlargest(10, "No_of_Votes")[["Series_Title", "Genre","IMDB_Rating","No_of_Votes"]]
```

```python
# problem -2 Hight Quality Movies Shortlist
# find movies with rating at least 8.and at least 100,000
shortlist = csv[(csv["IMDB_Rating"] >= 8.5) & (csv["No_of_Votes"] >= 100000)] 
shortlist[["Series_Title", "Genre", "IMDB_Rating", "No_of_Votes" ]]
```

```python
# Problem 3
# genre performane report
# which genre have hight rating longer runtime and more audice vote
# log, groupby, agg, max, mean, sum
genre_report = csv.groupby("Genre").agg(
    Movie_Count=("Series_Title", "count"),
    Average_Rating=("IMDB_Rating", "mean"),
    Median_Rating=("IMDB_Rating", "median"),
    Average_Runtime=("Runtime", "mean"),
    Total_Votes=("No_of_Votes", "sum")
)

genre_report.sort_values(
    "Average_Rating", ascending=False
)
```
